```python
import os
import pandas as pd
from sqlalchemy import create_engine, types  # <--- Added 'types' and 'create_engine'

def insert_into_db(path, db_name):
    dbs = {}
    all_files = os.listdir(path)
    all_files = [item for item in all_files if item.endswith(".csv")]
    
    for item in all_files:
        tab_name = item.split('.')[0]
        clean_path = os.path.join(path, item)
        dbs[tab_name] = pd.read_csv(clean_path)
        
    USERNAME = "root"
    PASSWORD = "python123"
    HOST = "localhost"
    PORT = "3306"
    DATABASE = db_name
    
    engine = create_engine(
        f"mysql+pymysql://{USERNAME}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}"
    )
    
    for item in dbs.keys():
        df = dbs[item]
        sql_types = {}
    
        if 'definition' in df.columns:
            sql_types['definition'] = types.Text() 
    
        df.to_sql(item, if_exists="replace", con=engine, index=False, dtype=sql_types)
            
    return "Inserted into database successfully"
```
