```python
data = {
    'State': ['Maharashtra', 'Maharashtra', 'Uttar Pradesh', 'Uttar Pradesh', 'Karnataka'],
    'Orders': [50, 80, 70, 90, 60],
    'Amount': [500, 800, 700, 900, 600]
}

temps = pd.DataFrame(data)

# Grouping by 'State' and summing the 'Orders'
grouped = temps.groupby('State').sum()

grouped
```

```python
# Grouping by 'State' and summing the 'Orders', keeping 'State' as a column
grouped = temps.groupby('State', as_index=False).sum()

grouped  # temps.groupby('State').sum()
```
